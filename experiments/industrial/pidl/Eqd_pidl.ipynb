```python
def make_par(s: str) -> str:
    store = []
    store_close = []
    items = []

    stack = []

    result = list(s)

    if len(result) % 2 == 0:

        for i in result:
            if i == '(':
                store.append(i)
            elif i == ')':
                store_close.append(i)
            else:
                items.append(i)
        
    else:
        print('invalid parentheses pairing')

    stack.extend(store)
    stack.extend(items)
    stack.extend(store_close)

    output = ''.join(stack)

    return output

        
```

```python
# Preprocessing Function
def preprocess_data(data):
    processed_data = data.copy()

    # Log-transform skewed features
    skewed_features = ["TMC_filtered", "CF_FL"]
    for feature in skewed_features:
        processed_data[feature] = np.log1p(processed_data[feature])

    # Binarize near-zero features
    binarize_features = ["BrakingTime_2", "disc_heating_BTM_FL"]
    for feature in binarize_features:
        processed_data[f"{feature}_binary"] = (processed_data[feature] > 0).astype(int)

    # Normalize all features to range [0, 1]
    for feature in data.columns:
        processed_data[f"{feature}_normalized"] = (
            (processed_data[feature] - processed_data[feature].min()) /
            (processed_data[feature].max() - processed_data[feature].min())
        )

    # Normalize target variable CF_FL after log transformation
    processed_data["CF_FL_normalized"] = (
        (processed_data["CF_FL"] - processed_data["CF_FL"].min()) /
        (processed_data["CF_FL"].max() - processed_data["CF_FL"].min())
    )

    return processed_data

# Derived Features
def create_derived_features(data):
    derived_data = data.copy()

    # Rolling averages (e.g., 5-sample window)
    derived_data["BrakingTime_2_rolling_mean"] = data["BrakingTime_2"].rolling(window=5, min_periods=1).mean()

    # Rate of change
    derived_data["TMC_filtered_rate"] = data["TMC_filtered"].diff().fillna(0)

    # Interaction terms
    derived_data["BrakingTime_POS_interaction"] = data["BrakingTime_2"] * data["POS_MEASUREREV"]

    return derived_data

# Apply Preprocessing
preprocessed_data = preprocess_data(data)

derived_data = create_derived_features(preprocessed_data)

# Display results
print("Preprocessed Data Sample:")
print(preprocessed_data.head())

print("Derived Features Sample:")
print(derived_data.head())
```

```python
from sklearn.ensemble import RandomForestRegressor
from sklearn.feature_selection import RFECV
from sklearn.model_selection import train_test_split

def feature_selection_with_rfecv(X, y, model):
    """
    Perform Recursive Feature Elimination with Cross-Validation (RFECV) to select important features.
    
    Parameters:
    - X: pd.DataFrame, Feature matrix
    - y: pd.Series, Continuous target variable
    - model: sklearn model, The model to use for RFECV (e.g., RandomForestRegressor)
    
    Returns:
    - selected_features: list, Names of the selected features
    - feature_ranking: list, Ranking of all features (1 means selected)
    - optimal_number_of_features: int, Optimal number of features selected by RFECV
    """
    
    # Initialize RFECV with cross-validation
    rfecv = RFECV(estimator=model, step=1, cv=5, scoring='neg_mean_squared_error', min_features_to_select=20)  
    
    # Fit RFECV
    rfecv.fit(X, y)
    
    # Get the selected features
    selected_features = list(X.columns[rfecv.support_])
    
    # Get the ranking of features
    feature_ranking = list(rfecv.ranking_)
    
    # Get the optimal number of features
    optimal_number_of_features = rfecv.n_features_
    
    return selected_features, feature_ranking, optimal_number_of_features

```

```python
from sklearn.ensemble import RandomForestRegressor
from sklearn.feature_selection import RFECV
from sklearn.model_selection import train_test_split

def feature_selection_with_rfecv(X, y, model):
    """
    Perform Recursive Feature Elimination with Cross-Validation (RFECV) to select important features.
    
    Parameters:
    - X: pd.DataFrame, Feature matrix
    - y: pd.Series, Continuous target variable
    - model: sklearn model, The model to use for RFECV (e.g., RandomForestRegressor)
    
    Returns:
    - selected_features: list, Names of the selected features
    - feature_ranking: list, Ranking of all features (1 means selected)
    - optimal_number_of_features: int, Optimal number of features selected by RFECV
    """
    
    # Initialize RFECV with cross-validation
    rfecv = RFECV(estimator=model, step=1, cv=5, scoring='neg_mean_squared_error', min_features_to_select=4)  
    
    # Fit RFECV
    rfecv.fit(X, y)
    
    # Get the selected features
    selected_features = list(X.columns[rfecv.support_])
    
    # Get the ranking of features
    feature_ranking = list(rfecv.ranking_)
    
    # Get the optimal number of features
    optimal_number_of_features = rfecv.n_features_
    
    return selected_features, feature_ranking, optimal_number_of_features

```
