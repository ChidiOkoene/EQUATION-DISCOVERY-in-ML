In order to discover PDEs in the data using symbolic regression, we need to compute spatial and temporal derivatives *u(x,t)*.
we can use either finite differences method or more accurately make use of Neural Networks to compute the derivatives.

**Using Finite Differences**

```python
# Calculate the spatial and temporal step sizes
dx = x[1] - x[0]  # Assuming uniform spacing
dt = t[1] - t[0]  # Assuming uniform spacing
```

```python
# First-order derivatives
u_x = np.gradient(usol, axis=1, edge_order=2) / dx  # du/dx
u_t = np.gradient(usol, axis=0, edge_order=2) / dt  # du/dt
```

```python
# Second-order spatial derivatives
u_xx = np.gradient(u_x, axis=1, edge_order=2) / dx  # d²u/dx²
```

```python
# Flatten the arrays and organize into a DataFrame
df_2 = pd.DataFrame({
    't': T.flatten(),
    'x': X.flatten(),
    'u': usol.flatten(),
    'u_t': u_t.flatten(),
    'u_x': u_x.flatten(),
    'u_xx': u_xx.flatten(),
})
```

```python
# Save the DataFrame for use in symbolic regression tools
df_2.to_csv("data/burgers_data_with_derivatives.csv", index=False)
```

```python
# Define features and interactions
features = ['u', 'u_x', 'u_xx']
target = df_2['u_t']  # Target variable
feature_matrix = pd.DataFrame()
```

```python
# Add raw features
for feature in features:
    feature_matrix[feature] = df_2[feature]
```

```python
# Add interaction terms
for degree in range(2, 4):  # Include up to degree-3 interactions
    for combo in combinations_with_replacement(features, degree):
        name = '*'.join(combo)
        term = df_2[combo[0]]
        for var in combo[1:]:
            term *= df_2[var]
        feature_matrix[name] = term
```

```python
from pysr import PySRRegressor
import numpy as np

# Define a function to calculate diversity in equations
def calculate_diversity(equations):
    """
    Measure the diversity of equations based on unique terms and structures.
    """
    if len(equations) < 2:
        return 0  # No diversity if there's only one equation

    equation_strings = [str(eq) for eq in equations]
    unique_equations = set(equation_strings)
    return len(unique_equations) / len(equation_strings)

# Adaptive PySR wrapper

def adaptive_pysr(features, target, initial_params, max_phases=3, iterations_per_phase=50):
    """
    Run PySR with adaptive crossover and mutation rates.

    Args:
        features: Input features (numpy array or pandas DataFrame).
        target: Target variable (numpy array or pandas Series).
        initial_params: Dict with initial parameters for crossover and mutation rates.
        max_phases: Number of adaptive phases.
        iterations_per_phase: Iterations per phase.

    Returns:
        Trained PySR model.
    """
    # Initialize PySR model
    model = PySRRegressor(
        niterations=iterations_per_phase,
        population_size=1000,
        operators=["+", "-", "*", "/", "sin", "cos", "exp", "log"],
        verbosity=1,
        progress=True,
        **initial_params
    )

    for phase in range(max_phases):
        print(f"\n=== Phase {phase + 1}/{max_phases} ===")

        # Fit the model for the current phase
        model.fit(features, target)

        # Evaluate diversity
        diversity = calculate_diversity(model.equations_["equation"])
        print(f"Diversity of equations: {diversity:.2f}")

        # Adjust crossover and mutation rates
        if diversity < 0.5:  # Low diversity, increase mutation
            model.mutation_probability = min(model.mutation_probability + 0.1, 0.6)
            print(f"Increased mutation rate to {model.mutation_probability}")
        else:  # High diversity, focus on exploitation
            model.crossover_probability = min(model.crossover_probability + 0.1, 0.8)
            print(f"Increased crossover rate to {model.crossover_probability}")

        # Increment iterations for the next phase
        model.niterations += iterations_per_phase

    return model

# Example usage:
# Load your dataset
# Replace with your own features and target
import pandas as pd

data = pd.read_csv("/mnt/data/clamping_force2_norm.csv")
features = data.drop(columns=["CF_FL"])  # Replace with actual target column name
target = data["CF_FL"]

# Initial parameters
initial_params = {
    "crossover_probability": 0.5,
    "mutation_probability": 0.2,
}

# Run adaptive PySR
model = adaptive_pysr(features, target, initial_params, max_phases=3, iterations_per_phase=50)

# View final equations
print(model.equations_)
```
